```python

# ============================================================
# NETFLIX DATA ANALYSIS AND CONTENT RECOMMENDATION SYSTEM
# DAES MINI PROJECT - GOOGLE COLAB
# ============================================================

# SESSION 1: IMPORT LIBRARIES

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    mean_squared_error, r2_score,
    accuracy_score, confusion_matrix,
    classification_report, silhouette_score
)

from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from scipy.stats import zscore
from scipy.cluster.hierarchy import linkage, dendrogram

print("All libraries imported successfully!")


# ============================================================
# DATASET UPLOAD AND LOADING
# ============================================================

print("Upload netflix_titles.csv")

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
display(df.head())


# ============================================================
# SESSION 1: NUMPY AND PANDAS BASICS
# ============================================================

print("\nDATASET COLUMNS:")
print(df.columns.tolist())

print("\nDATASET INFORMATION:")
df.info()

print("\nSTATISTICAL DESCRIPTION:")
display(df.describe(include='all'))

print("\nMOVIE AND TV SHOW COUNTS:")
print(df['type'].value_counts())


# ============================================================
# SESSION 2: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

print("\nMISSING VALUES:")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS:")
print(df.duplicated().sum())

print("\nCONTENT TYPE DISTRIBUTION:")
print(df['type'].value_counts())

print("\nTOP 10 COUNTRIES:")
print(df['country'].dropna().str.split(',').explode().str.strip().value_counts().head(10))

print("\nTOP 10 DIRECTORS:")
print(df['director'].dropna().str.split(',').explode().str.strip().value_counts().head(10))

print("\nTOP 10 GENRES:")
print(df['listed_in'].dropna().str.split(',').explode().str.strip().value_counts().head(10))


# ============================================================
# SESSION 3: DATA VISUALIZATION
# ============================================================

# 1. Movies vs TV Shows

plt.figure(figsize=(7,5))
sns.countplot(data=df, x='type')
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()


# 2. Content released by year

plt.figure(figsize=(12,5))
df['release_year'].value_counts().sort_index().plot()
plt.title('Netflix Content Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()


# 3. Top 10 countries

countries = df['country'].dropna().str.split(',').explode().str.strip()

plt.figure(figsize=(10,5))
countries.value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()


# 4. Top 10 genres

genres = df['listed_in'].dropna().str.split(',').explode().str.strip()

plt.figure(figsize=(10,5))
genres.value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()


# 5. Release year histogram

plt.figure(figsize=(9,5))
sns.histplot(df['release_year'].dropna(), bins=30, kde=True)
plt.title('Distribution of Release Years')
plt.xlabel('Release Year')
plt.show()


# 6. Box plot

plt.figure(figsize=(8,4))
sns.boxplot(x=df['release_year'])
plt.title('Release Year Outlier Visualization')
plt.show()


# 7. Scatter plot

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='release_year', y=df['release_year'].value_counts().reindex(df['release_year']).values)
plt.title('Release Year Distribution')
plt.xlabel('Release Year')
plt.ylabel('Frequency')
plt.show()


# 8. Correlation heatmap

plt.figure(figsize=(7,5))
sns.heatmap(df[['release_year']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()


# =====================================================
# NETFLIX CONTENT-BASED RECOMMENDATION SYSTEM
# =====================================================

# STEP 1: IMPORT LIBRARIES

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")


# STEP 2: LOAD DATASET

# If df is already loaded in your previous code,
# this step is not required.

# Otherwise, uncomment these two lines:
# from google.colab import files
# uploaded = files.upload()
# df = pd.read_csv('netflix_titles.csv')

# Create a separate copy
netflix = df.copy()

print("Dataset loaded successfully!")
print("Total records:", len(netflix))


# STEP 3: DATA PREPROCESSING

# Handle missing values

netflix['title'] = netflix['title'].fillna('')
netflix['listed_in'] = netflix['listed_in'].fillna('')
netflix['description'] = netflix['description'].fillna('')
netflix['director'] = netflix['director'].fillna('')
netflix['cast'] = netflix['cast'].fillna('')
netflix['type'] = netflix['type'].fillna('')

# Remove duplicate records

netflix = netflix.drop_duplicates(subset='show_id')

# Reset index

netflix = netflix.reset_index(drop=True)

print("Data preprocessing completed!")


# STEP 4: FEATURE ENGINEERING

# Combine important content features

netflix['combined_features'] = (
    netflix['listed_in'] + ' ' +
    netflix['description'] + ' ' +
    netflix['director'] + ' ' +
    netflix['cast'] + ' ' +
    netflix['type']
)

print("\nCombined features created!")

display(netflix[
    ['title', 'combined_features']
].head())


# STEP 5: TF-IDF VECTORIZATION

# Convert text into numerical vectors

tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=5000
)

tfidf_matrix = tfidf.fit_transform(
    netflix['combined_features']
)

print("\nTF-IDF vectorization completed!")

print("TF-IDF matrix shape:", tfidf_matrix.shape)


# STEP 6: CREATE RECOMMENDATION FUNCTION

def recommend_content(movie_name, number=5):

    # Search for the selected movie or show

    matches = netflix[
        netflix['title'].str.lower() ==
        movie_name.strip().lower()
    ]

    # If exact title is not found, search partially

    if matches.empty:

        matches = netflix[
            netflix['title'].str.contains(
                movie_name,
                case=False,
                na=False,
                regex=False
            )
        ]

    # Check whether title exists

    if matches.empty:

        print("Sorry! Title not found.")
        print("Please enter another movie or TV show name.")
        return

    # Get the first matching title

    index = matches.index[0]

    # Calculate similarity with all titles

    similarity_scores = cosine_similarity(
        tfidf_matrix[index],
        tfidf_matrix
    ).flatten()

    # Exclude the selected title

    similarity_scores[index] = -1

    # Get top similar titles

    top_indices = np.argsort(
        similarity_scores
    )[::-1][:number]

    # Display selected title

    print("\n================================")
    print("SELECTED NETFLIX CONTENT")
    print("================================")

    print("Title:", netflix.loc[index, 'title'])
    print("Type:", netflix.loc[index, 'type'])
    print("Genre:", netflix.loc[index, 'listed_in'])

    # Display recommendations

    print("\n================================")
    print("RECOMMENDED CONTENT")
    print("================================")

    recommendations = netflix.iloc[
        top_indices
    ][
        ['title', 'type', 'release_year',
         'rating', 'listed_in']
    ].copy()

    recommendations['Similarity (%)'] = (
        similarity_scores[top_indices] * 100
    ).round(2)

    display(recommendations.reset_index(drop=True))


# STEP 7: TEST RECOMMENDATION SYSTEM

recommend_content('Stranger Things', 5)


# STEP 8: ALLOW USER TO ENTER A TITLE

print("\n================================")
print("NETFLIX RECOMMENDATION SYSTEM")
print("================================")

movie_input = input(
    "Enter a Netflix movie or TV show name: "
)

recommend_content(movie_input, 5)


# STEP 9: DISPLAY RECOMMENDATION VISUALIZATION

def plot_recommendations(movie_name):

    matches = netflix[
        netflix['title'].str.lower() ==
        movie_name.strip().lower()
    ]

    if matches.empty:
        matches = netflix[
            netflix['title'].str.contains(
                movie_name,
                case=False,
                na=False,
                regex=False
            )
        ]

    if matches.empty:
        print("Title not found for visualization.")
        return

    index = matches.index[0]

    similarity_scores = cosine_similarity(
        tfidf_matrix[index],
        tfidf_matrix
    ).flatten()

    similarity_scores[index] = -1

    top_indices = np.argsort(
        similarity_scores
    )[::-1][:5]

    result = netflix.iloc[top_indices]

    plt.figure(figsize=(10, 5))

    plt.barh(
        result['title'].values[::-1],
        (similarity_scores[top_indices] * 100)[::-1]
    )

    plt.xlabel("Similarity Percentage")
    plt.ylabel("Recommended Titles")

    plt.title(
        "Top 5 Netflix Recommendations for "
        + netflix.loc[index, 'title']
    )

    plt.tight_layout()
    plt.show()


# STEP 10: SHOW RECOMMENDATION GRAPH

plot_recommendations('Stranger Things')


# =====================================================
# FINAL MESSAGE
# =====================================================

print("\n================================")
print("PROJECT COMPLETED SUCCESSFULLY!")
print("================================")

print("Recommendation Algorithm: TF-IDF")
print("Similarity Algorithm: Cosine Similarity")
print("Recommendation Type: Content-Based Filtering")
print("Number of Recommendations: 5")
```
